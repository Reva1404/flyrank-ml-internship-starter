## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

This lane is a scoring task. The goal is to assign each page a Refresh Score and rank pages so content teams know which ones should be reviewed first for refresh, expansion, protection, pruning, or monitoring.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

The target is a Refresh Score (or refresh priority) that ranks pages by how urgently they should be reviewed.

Since the dataset does not contain product decision scores or refresh labels, the model uses a defined proxy built from observable SEO signals such as declining trends, impressions, CTR, engagement, freshness, and position to estimate refresh priority. The objective is to produce a ranked review queue rather than predict a guaranteed outcome.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

Precision@K (Precision@50)

The primary metric is Precision@50, which measures how many of the top 50 pages recommended for review are truly good refresh candidates. This metric aligns with the real-world workflow because reviewers can only inspect a limited number of pages at a time. Average Precision can also be reported as a secondary metric

## 4. The unit of analysis, as a real dataframe

*Load your lane's slice and show it: one row = one what?*

Each row in the dataframe represents one content page (one webpage/content item) together with its observed SEO, engagement, and content features.

Example columns include:

content_id,
client_id,
impressions_90d,
clicks_90d,
sessions_90d,
content_age_days,
ctr,
avg_position,
engagement_rate,
trend_direction

One row = one content page being evaluated for refresh priority.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A fixed rule can only evaluate one or two conditions at a time, such as content age or declining impressions. Refresh priority, however, depends on multiple interacting signals including visibility, freshness, CTR, average position, engagement, and historical trends. Machine learning can learn these complex relationships and generate a ranked review queue with confidence scores and reason codes, providing better decision support than a transparent rule alone while still requiring human review.
